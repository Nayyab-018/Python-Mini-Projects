```python
import random

def play_cyber_vault():
    secret_code = random.randint(1, 25)
    attempts = 0
    max_attempts = 5

    print("🔐 === CYBER-VAULT SECURITY CRACKER ===")
    print("A secret 2-digit security code (1 to 25) is protecting the mainframe.")
    print(f"You have {max_attempts} attempts to crack it before lockdown!\n")

    while attempts < max_attempts:
        try:
            guess = int(input(f"Attempt {attempts + 1}/{max_attempts} — Enter code: "))
        except ValueError:
            print("⚠️ Invalid input! Please enter a whole number between 1 and 25.\n")
            continue

        if guess < 1 or guess > 25:
            print("⚠️ Out of range! Enter a number between 1 and 25.\n")
            continue

        attempts += 1

        if guess == secret_code:
            score = (max_attempts - attempts + 1) * 20
            print(f"\n🎉 ACCESS GRANTED! You cracked the vault code ({secret_code}) in {attempts} tries!")
            print(f"🏆 Security Score: {score}/100")
            break
        elif guess < secret_code:
            print("📉 Too low! Aim higher.\n")
        else:
            print("📈 Too high! Aim lower.\n")
    else:
        print(f"\n🚨 SYSTEM LOCKDOWN! Out of attempts. The secret code was {secret_code}.")

play_cyber_vault()
```
