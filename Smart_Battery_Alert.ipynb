```python
def check_battery(percent, is_charging):
    if percent < 0 or percent > 100:
        return "❌ Invalid battery percentage! Please enter 0 to 100."
    elif percent <= 20 and not is_charging:
        return "🪫 LOW BATTERY: Plug in your charger now to prevent shutdown!"
    elif percent >= 80 and is_charging:
        return "⚡ OPTIMAL LIMIT REACHED: Unplug charger to protect battery health!"
    elif percent == 100 and is_charging:
        return "🚨 OVERCHARGING WARNING: Battery is full. Disconnect power immediately!"
    elif is_charging:
        return f"🔌 Charging normally ({percent}%). Keep charging until 80%."
    else:
        return f"✅ Battery healthy at {percent}%. No action needed."

print("--- SMART PHONE CHARGING ASSISTANT ---")
battery_level = int(input("Enter current battery percentage (0-100): "))
charger_status = input("Is the charger plugged in? (yes/no): ").strip().lower()

charging = (charger_status == "yes")
advice = check_battery(battery_level, charging)
print("\n" + advice)
```
